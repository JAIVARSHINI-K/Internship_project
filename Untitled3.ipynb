```python
import numpy as np
import random

# Number of traffic states (0 to 9 vehicles)
states = 10

# Actions:
# 0 = RED
# 1 = GREEN
actions = 2

# Q-Table
Q = np.zeros((states, actions))

# Parameters
alpha = 0.1      # Learning Rate
gamma = 0.9      # Discount Factor
epsilon = 0.2    # Exploration Rate
episodes = 1000

# Training
for episode in range(episodes):

    traffic = random.randint(0, 9)

    for step in range(20):

        # Choose action
        if random.uniform(0, 1) < epsilon:
            action = random.randint(0, 1)
        else:
            action = np.argmax(Q[traffic])

        # Reward Logic
        if traffic >= 5:
            reward_green = 10
            reward_red = -10
        else:
            reward_green = -2
            reward_red = 5

        # Apply action
        if action == 1:  # GREEN
            reward = reward_green
            next_traffic = max(0, traffic - random.randint(1, 3))
        else:            # RED
            reward = reward_red
            next_traffic = min(9, traffic + random.randint(0, 2))

        # Q-Learning Update
        old_value = Q[traffic, action]
        next_max = np.max(Q[next_traffic])

        Q[traffic, action] = old_value + alpha * (
            reward + gamma * next_max - old_value
        )

        traffic = next_traffic

# Display Q-Table
print("Learned Q-Table:\n")
print(Q)

# Final Decisions
print("\nTraffic Signal Decisions")

for state in range(states):

    action = np.argmax(Q[state])

    if action == 1:
        decision = "GREEN"
    else:
        decision = "RED"

    print(f"Vehicles Waiting: {state} --> Signal: {decision}")
```
